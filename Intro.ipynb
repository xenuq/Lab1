```python
#Написать функцию на вход которой подается строка, состоящая из латинских букв. Функция должна вернуть количество гласных букв (a, e, i, o, u) в этой строке. Написать тесты для кода
```

```python
def count_vowels(s):
    vowels = "aeiou"
    count = 0
    for char in s.lower():
        if char in vowels:
            count += 1
    return count

# Тесты
def test_count_vowels():
    # Тест 1: Пустая строка
    assert count_vowels("") == 0, "Пустая строка должна возвращать 0"
    
    # Тест 2: Строка без гласных
    assert count_vowels("bcdfg") == 0, "Строка без гласных должна возвращать 0"
    
    # Тест 3: Строка только с гласными
    assert count_vowels("aeiou") == 5, "Строка 'aeiou' должна возвращать 5"
    
    # Тест 4: Строка с гласными в верхнем регистре
    assert count_vowels("AEIOU") == 5, "Строка 'AEIOU' должна возвращать 5"
    
    # Тест 5: Строка со смешанными регистрами
    assert count_vowels("aEiOu") == 5, "Строка 'aEiOu' должна возвращать 5"
    
    # Тест 6: Обычная строка с гласными и согласными
    assert count_vowels("hello world") == 3, "Строка 'hello world' должна возвращать 3"
    
    # Тест 7: Строка с повторяющимися гласными
    assert count_vowels("programming") == 3, "Строка 'programming' должна возвращать 3"
    
    # Тест 8: Строка с пробелами и знаками препинания
    assert count_vowels("Hello, World!") == 3, "Строка 'Hello, World!' должна возвращать 3"
    
    print("Все тесты пройдены успешно!")

if __name__ == "__main__":
    test_count_vowels()
```

```python
#Написать функцию на вход, которой подается строка. Функция должна вернуть true, если каждый символ в строке встречается только 1 раз, иначе должна вернуть false. Написать тесты для кода
```

```python
def has_unique_chars(s):
    return len(s) == len(set(s))

# Тесты
def test_has_unique_chars():
    # Тест 1: Пустая строка
    assert has_unique_chars("") == True, "Пустая строка должна возвращать True"
    
    # Тест 2: Все символы уникальны
    assert has_unique_chars("abcde") == True, "Строка 'abcde' должна возвращать True"
    
    # Тест 3: Есть повторяющиеся символы
    assert has_unique_chars("hello") == False, "Строка 'hello' должна возвращать False"
    
    # Тест 4: Один символ
    assert has_unique_chars("a") == True, "Строка 'a' должна возвращать True"
    
    # Тест 5: Два одинаковых символа
    assert has_unique_chars("aa") == False, "Строка 'aa' должна возвращать False"
    
    # Тест 6: Символы в разном регистре
    assert has_unique_chars("aA") == True, "Строка 'aA' должна возвращать True (разные регистры)"
    
    # Тест 7: Строка с пробелами (уникальные символы)
    assert has_unique_chars("a b") == True, "Строка 'a b' должна возвращать True"
    
    # Тест 8: Строка с повторяющимися пробелами
    assert has_unique_chars("a  b") == False, "Строка 'a  b' должна возвращать False"
    
    # Тест 9: Специальные символы
    assert has_unique_chars("!@#$%") == True, "Строка '!@#$%' должна возвращать True"
    
    # Тест 10: Длинная строка с повторениями
    assert has_unique_chars("abcdefghijklmnopqrstuvwxyz") == True, "Алфавит должен возвращать True"
    assert has_unique_chars("abcdefghijklmnopqrstuvwxyza") == False, "Алфавит + 'a' должен возвращать False"
    
    print("Все тесты пройдены успешно!")

# Запуск тестов
if __name__ == "__main__":
    test_has_unique_chars()
```

```python
def count_bits(n):
    return bin(n).count('1')

# Тесты
def test_count_bits():
    # Тест 1: Число 0
    assert count_bits(0) == 0, "Число 0 должно возвращать 0"
    
    # Тест 2: Число 1
    assert count_bits(1) == 1, "Число 1 должно возвращать 1"
    
    # Тест 3: Число 2
    assert count_bits(2) == 1, "Число 2 должно возвращать 1"
    
    # Тест 4: Число 3
    assert count_bits(3) == 2, "Число 3 должно возвращать 2"
    
    # Тест 5: Число 7 (111 в двоичной)
    assert count_bits(7) == 3, "Число 7 должно возвращать 3"
    
    # Тест 6: Число 10 (1010 в двоичной)
    assert count_bits(10) == 2, "Число 10 должно возвращать 2"
    
    # Тест 7: Число 15 (1111 в двоичной)
    assert count_bits(15) == 4, "Число 15 должно возвращать 4"
    
    # Тест 8: Число 255 (все биты равны 1)
    assert count_bits(255) == 8, "Число 255 должно возвращать 8"
    
    # Тест 9: Большое число
    assert count_bits(1024) == 1, "Число 1024 должно возвращать 1"
    
    # Тест 10: Число 1023 (все биты равны 1 для 10 бит)
    assert count_bits(1023) == 10, "Число 1023 должно возвращать 10"
    
    print("Все тесты пройдены успешно!")

if __name__ == "__main__":
    test_count_bits()
```

```python
#Написать функцию, которая принимает положительное число. Функция должна вернуть то, сколько раз необходимо перемножать цифры числа или результат перемножения, чтобы получилось число состоящее из одной цифры. Например, для входного числа: · 39 функция должна вернуть 3, так как 39=27 => 27=14 => 14=4 · 4 функция должна вернуть 0, так как число уже состоит из одной цифры · 999 функция должна вернуть 4, так как 999=729 => 729=126 => 126=12 => 12=2. Написать тесты для кода
```

```python
def persistence(n):
    count = 0
    while n >= 10:
        product = 1
        for digit in str(n):
            product *= int(digit)
        n = product
        count += 1
    return count

# Тесты
def test_persistence():
    # Тест 1: Число из одной цифры
    assert persistence(4) == 0, "Число 4 должно возвращать 0"
    
    # Тест 2: Пример из задания
    assert persistence(39) == 3, "Число 39 должно возвращать 3"
    
    # Тест 3: Другой пример из задания
    assert persistence(999) == 4, "Число 999 должно возвращать 4"
    
    # Тест 4: Число 10
    assert persistence(10) == 1, "Число 10 должно возвращать 1 (1*0=0)"
    
    # Тест 5: Число 25
    assert persistence(25) == 2, "Число 25 должно возвращать 2 (2*5=10 => 1*0=0)"
    
    # Тест 6: Число 1
    assert persistence(1) == 0, "Число 1 должно возвращать 0"
    
    # Тест 7: Число 0
    assert persistence(0) == 0, "Число 0 должно возвращать 0"
    
    # Тест 8: Число 123
    assert persistence(123) == 1, "Число 123 должно возвращать 1 (1*2*3=6)"
    
    # Тест 9: Число 6788
    assert persistence(6788) == 6, "Число 6788 должно возвращать 6"
    
    # Тест 10: Число 77
    assert persistence(77) == 4, "Число 77 должно возвращать 4 (7*7=49 => 4*9=36 => 3*6=18 => 1*8=8)"
    
    print("Все тесты пройдены успешно!")

if __name__ == "__main__":
    test_persistence()
```

```python
#Написать функцию, которая принимает два целочисленных вектора одинаковой длины и возвращает среднеквадратическое отклонение двух векторов. Написать тесты для кода
```

```python
def mse(vector1, vector2):
    if len(vector1) != len(vector2):
        raise ValueError("Векторы должны быть одинаковой длины")
    
    if len(vector1) == 0:
        return 0
    
    squared_errors = [(v1 - v2) ** 2 for v1, v2 in zip(vector1, vector2)]
    return sum(squared_errors) / len(vector1)

# Тесты
def test_mse():
    # Тест 1: Одинаковые векторы
    assert mse([1, 2, 3], [1, 2, 3]) == 0, "Одинаковые векторы должны давать MSE = 0"
    
    # Тест 2: Векторы с постоянной разницей
    assert mse([1, 2, 3], [2, 3, 4]) == 1, "Векторы с разницей 1 должны давать MSE = 1"
    
    # Тест 3: Векторы с разницей 2
    assert mse([1, 2, 3], [3, 4, 5]) == 4, "Векторы с разницей 2 должны давать MSE = 4"
    
    # Тест 4: Один элемент
    assert mse([5], [3]) == 4, "Векторы [5] и [3] должны давать MSE = 4"
    
    # Тест 5: Отрицательные числа
    result = mse([-1, -2, -3], [1, 2, 3])
    expected = 56 / 3
    assert abs(result - expected) < 1e-10, f"Ожидалось {expected}, получено {result}"
    
    # Тест 6: Пустые векторы
    assert mse([], []) == 0, "Пустые векторы должны давать MSE = 0"
    
    # Тест 7: Дробные числа (ИСПРАВЛЕНО)
    # [1.5, 2.5] и [1.0, 2.0]: разницы: [0.5, 0.5], квадраты: [0.25, 0.25], сумма: 0.5, MSE: 0.5/2 = 0.25
    result = mse([1.5, 2.5], [1.0, 2.0])
    expected = 0.25
    assert abs(result - expected) < 1e-10, f"Ожидалось {expected}, получено {result}"
    
    # Тест 8: Разные значения ошибок
    result = mse([1, 2, 3], [4, 5, 6])
    expected = ((1-4)**2 + (2-5)**2 + (3-6)**2) / 3
    assert abs(result - expected) < 1e-10, "Ручной расчет должен совпадать с функцией"
    
    # Тест 9: Проверка на исключение при разной длине
    try:
        mse([1, 2], [1, 2, 3])
        assert False, "Должно быть вызвано исключение для векторов разной длины"
    except ValueError:
        pass  # Ожидаемое поведение
    
    print("Все тесты пройдены успешно!")

if __name__ == "__main__":
    test_mse()
```

```python
#Написать функцию, принимающая целое положительное число. Функция должна вернуть строку вида “(n1p1)(n2p2)…(nkpk)” представляющая разложение числа на простые множители (если pi == 1, то выводить только ni). Например, для числа 86240 функция должна вернуть “(25)(5)(7**2)(11)”. Написать тесты для кода
```

```python
def prime_factors(n):
    if n == 1:
        return "(1)"
    
    factors = {}
    d = 2
    while d * d <= n:
        while n % d == 0:
            factors[d] = factors.get(d, 0) + 1
            n //= d
        d += 1
    if n > 1:
        factors[n] = factors.get(n, 0) + 1
    
    result = []
    for prime in sorted(factors.keys()):
        power = factors[prime]
        if power == 1:
            result.append(f"({prime})")
        else:
            result.append(f"({prime}**{power})")
    
    return "".join(result)

# Тесты
def test_prime_factors():
    # Тест 1: Число 1
    assert prime_factors(1) == "(1)", "Число 1 должно возвращать (1)"
    
    # Тест 2: Простое число
    assert prime_factors(7) == "(7)", "Простое число 7 должно возвращать (7)"
    
    # Тест 3: Число из задания
    assert prime_factors(86240) == "(2**5)(5)(7**2)(11)", "Число 86240 должно возвращать (2**5)(5)(7**2)(11)"
    
    # Тест 4: Число 2
    assert prime_factors(2) == "(2)", "Число 2 должно возвращать (2)"
    
    # Тест 5: Число 4
    assert prime_factors(4) == "(2**2)", "Число 4 должно возвращать (2**2)"
    
    # Тест 6: Число 6
    assert prime_factors(6) == "(2)(3)", "Число 6 должно возвращать (2)(3)"
    
    # Тест 7: Число 12
    assert prime_factors(12) == "(2**2)(3)", "Число 12 должно возвращать (2**2)(3)"
    
    # Тест 8: Число 100
    assert prime_factors(100) == "(2**2)(5**2)", "Число 100 должно возвращать (2**2)(5**2)"
    
    # Тест 9: Число 30
    assert prime_factors(30) == "(2)(3)(5)", "Число 30 должно возвращать (2)(3)(5)"
    
    # Тест 10: Большое число
    assert prime_factors(2310) == "(2)(3)(5)(7)(11)", "Число 2310 должно возвращать (2)(3)(5)(7)(11)"
    
    print("Все тесты пройдены успешно!")

if __name__ == "__main__":
    test_prime_factors()
```

```python
#Написать функцию, принимающая целое число n, задающее количество кубиков. Функция должна определить, можно ли из данного кол-ва кубиков построить пирамиду, то есть можно ли представить число n как 1^2+2^2+3^2+…+k^2. Если можно, то функция должна вернуть k, иначе строку “It is impossible”. Написать тесты для кода
```

```python
def pyramid(n):
    total = 0
    k = 1
    while total < n:
        total += k * k
        if total == n:
            return k
        k += 1
    return "It is impossible"

# Тесты
def test_pyramid():
    # Тест 1: Пирамида из 1 кубика
    assert pyramid(1) == 1, "1 кубик должен возвращать 1"
    
    # Тест 2: Пирамида из 5 кубиков (1² + 2² = 1 + 4)
    assert pyramid(5) == 2, "5 кубиков должны возвращать 2"
    
    # Тест 3: Пирамида из 14 кубиков (1² + 2² + 3² = 1 + 4 + 9)
    assert pyramid(14) == 3, "14 кубиков должны возвращать 3"
    
    # Тест 4: Пирамида из 30 кубиков (1² + 2² + 3² + 4² = 1 + 4 + 9 + 16)
    assert pyramid(30) == 4, "30 кубиков должны возвращать 4"
    
    # Тест 5: Пирамида из 55 кубиков (1² + 2² + 3² + 4² + 5² = 1 + 4 + 9 + 16 + 25)
    assert pyramid(55) == 5, "55 кубиков должны возвращать 5"
    
    # Тест 6: Невозможная пирамида
    assert pyramid(2) == "It is impossible", "2 кубика должны возвращать 'It is impossible'"
    
    # Тест 7: Невозможная пирамида
    assert pyramid(3) == "It is impossible", "3 кубика должны возвращать 'It is impossible'"
    
    # Тест 8: Невозможная пирамида
    assert pyramid(7) == "It is impossible", "7 кубиков должны возвращать 'It is impossible'"
    
    # Тест 9: Невозможная пирамида
    assert pyramid(15) == "It is impossible", "15 кубиков должны возвращать 'It is impossible'"
    
    # Тест 10: Большая пирамида
    assert pyramid(91) == 6, "91 кубик должен возвращать 6 (1+4+9+16+25+36)"
    
    print("Все тесты пройдены успешно!")

if __name__ == "__main__":
    test_pyramid()
```

```python
#Функция принимает на вход положительное число и определяет является ли оно сбалансированным, т.е. сумма цифр до средних равна сумме цифр после. Средними в случае нечетного числа цифр считать одну цифру, в случае четного - две средних. Написать тесты для кода
```

```python
def is_balanced(n):
    s = str(n)
    length = len(s)
    
    if length <= 1:
        return True
    
    if length % 2 == 0:
        mid = length // 2
        left_sum = sum(int(digit) for digit in s[:mid-1])
        right_sum = sum(int(digit) for digit in s[mid+1:])
    else:
        mid = length // 2
        left_sum = sum(int(digit) for digit in s[:mid])
        right_sum = sum(int(digit) for digit in s[mid+1:])
    
    return left_sum == right_sum

# Тесты
def test_is_balanced():
    # Тест 1: Однозначное число
    assert is_balanced(5) == True, "Однозначное число должно быть сбалансированным"
    
    # Тест 2: Двузначное число (обе цифры средние, поэтому суммы слева и справа = 0)
    assert is_balanced(12) == True, "Двузначное число должно быть сбалансированным (0 == 0)"
    
    # Тест 3: Сбалансированное трехзначное число
    assert is_balanced(121) == True, "121 должно быть сбалансированным (1 == 1)"
    
    # Тест 4: Несбалансированное трехзначное число
    assert is_balanced(123) == False, "123 должно быть несбалансированным (1 != 3)"
    
    # Тест 5: Сбалансированное четырехзначное число
    assert is_balanced(1221) == True, "1221 должно быть сбалансированным (1 == 1)"
    
    # Тест 6: Несбалансированное четырехзначное число
    assert is_balanced(1234) == False, "1234 должно быть несбалансированным (1 != 4)"
    
    # Тест 7: Сбалансированное пятизначное число
    assert is_balanced(12321) == True, "12321 должно быть сбалансированным (12 == 21)"
    
    # Тест 8: Несбалансированное пятизначное число
    assert is_balanced(12345) == False, "12345 должно быть несбалансированным (12 != 45)"
    
    # Тест 9: Сбалансированное шестизначное число
    assert is_balanced(123321) == True, "123321 должно быть сбалансированным (12 == 21)"
    
    # Тест 10: Несбалансированное шестизначное число
    assert is_balanced(123456) == False, "123456 должно быть несбалансированным (12 != 56)"
    
    # Тест 11: Число 0
    assert is_balanced(0) == True, "0 должно быть сбалансированным"
    
    # Тест 12: Большое сбалансированное число
    assert is_balanced(123404321) == True, "123404321 должно быть сбалансированным"
    
    print("Все тесты пройдены успешно!")

if __name__ == "__main__":
    test_is_balanced()
```
